#Data cleaning

```python
#Cắt bỏ ký tự $ và ép kiểu float
df_products['Product_Cost']= df_products['Product_Cost'].str.replace('$','').astype(float)
df_products['Product_Price']= df_products['Product_Price'].str.replace('$','').astype(float)

#Kiểm tra
df_products.info()
df_products.head()
```

```python
print("\nThống kê Products(Cost & Price)")
print(df_products[['Product_Cost', 'Product_Price']].describe())
```

- 2 cột Product_Cost và Product_Price đều mang định dạng số, không có số âm

```python
# Ép kiểu dữ liệu sang DateTime
df_sales['Date'] = pd.to_datetime(df_sales['Date'])
df_stores['Store_Open_Date'] = pd.to_datetime(df_stores['Store_Open_Date'])
df_sales.info()
df_stores.info()
```

###EDA

```python
#Kiểm tra liên kết Sales với Products
test_sales_prod = pd.merge(df_sales, df_products, on='Product_ID', how = 'left',indicator=True)
orphans_prod = test_sales_prod[test_sales_prod['_merge'] == 'left_only']
print(f'Số giao dịch Sales không tìm thấy Product_ID trong df_products: {len(orphans_prod)}')
```

```python
#Kiểm tra liên kết Sales với Stores
test_sales_stores = pd.merge(df_sales, df_stores, on = 'Store_ID', how = 'left', indicator=True)
orphans_stores = test_sales_stores[test_sales_stores['_merge'] == 'left_only']
print(f'Số giao dịch Sales không tìm thấy Store_ID trong df_stores: {len(orphans_stores)}')
```

```python
#Kiểm tra liên kết Inventory với Products
test_inventory_prod = pd.merge(df_inventory, df_products, on='Product_ID', how = 'left',indicator=True)
orphans_prod = test_inventory_prod[test_inventory_prod['_merge'] == 'left_only']
print(f'Số giao dịch Inventory không tìm thấy Product_ID trong df_products: {len(orphans_prod)}')
```

- Nhận xét: sau khi kiểm tra không thấy có dữ liệu chỉ có ở bảng bên trái(left_only) nên có thể dùng inner join để gộp bảng mà không bị mất dữ liệu

```python
#Gộp bảng Sales và Products, tạo 2 biến Revenue và Profit
df_total = pd.merge(df_sales, df_products, on = 'Product_ID', how = 'inner')
df_total['Revenue'] = df_total['Units'] * df_total['Product_Price']
df_total['Profit'] = df_total['Revenue'] - df_total['Units'] * df_total['Product_Cost']
df_total.head()
```

```python
df_total.info()
```

```python
#Tạo các biến thời gian
df_total['Month'] = df_total['Date'].dt.month
df_total['Day_of_Week'] = df_total['Date'].dt.day_name()
df_total['Year'] = df_total['Date'].dt.year
df_total['Quarter'] = df_total['Date'].dt.quarter
df_total[['Date', 'Month', 'Day_of_Week','Quarter', 'Year', 'Revenue','Profit']].head()
```

```python
#Lợi nhuận (Profit) theo danh mục sản phẩm (Product_Category)
category_profit = df_total.groupby('Product_Category')['Profit'].sum().sort_values(ascending=False)
category_profit.head()
```

```python
category_profit.plot(kind='barh', color='blue')
plt.title('Lợi nhuận theo danh mục sản phẩm')
plt.xlabel('Lợi nhuận')
plt.ylabel('Danh mục sản phẩm')
plt.ticklabel_format(style='plain', axis='x')
plt.show()
```

- Nhận xét:
- Nhóm Toys và Electronics mang về lợi nhuận áp đảo (vượt mốc 1 triệu đô)
- Mảng Sports & Outdoors lại đang chật vật ở chót bảng (chỉ bằng một nửa)


```python
#Gộp df_total với bảng Stores
df_total = pd.merge(df_total,df_stores, on = 'Store_ID', how = 'inner')
#Doanh thu (Revenue) theo vị trí cửa hàng
location_revenue = df_total.groupby('Store_Location')['Revenue'].sum().sort_values(ascending=False)
location_revenue
```

```python
location_revenue.plot(kind = 'bar', color = 'blue')
plt.title('Doanh thu theo vị trí cửa hàng')
plt.xlabel('Vị trí cửa hàng')
plt.ylabel('Doanh thu')
plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

- Nhận xét: Các cửa hàng ở khu vực Downtown (Trung tâm thành phố) mang về doanh thu áp đảo (hơn 8 triệu USD)

```python
trend_month = df_total.groupby('Month')['Revenue'].sum().sort_index()
trend_month
```

```python
trend_month.plot(kind='line',color = 'blue')
plt.title('Doanh thu theo tháng')
plt.xlabel('Tháng')
plt.ylabel('Tổng Doanh Thu ($)')
plt.xticks(range(1, 13))
plt.ticklabel_format(style='plain', axis='y')
plt.grid(True)
plt.show()
```

- Doanh thu cao rơi vào tháng 3 đến tháng 6 (cao nhất vào tháng 4)

```python


# Bước 1: Chỉ lọc ra các cột chứa con số để tính toán (tránh lỗi khi đưa chữ vào tính tương quan)
numeric_columns = df_total.select_dtypes(include=['float64', 'int64'])

# Bước 2: Bật Radar quét toàn bộ (tạo Correlation Matrix)
corr_matrix = numeric_columns.corr()

# Bước 3: Vẽ Bản đồ nhiệt (Heatmap) để sếp cũng hiểu được
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

plt.title('BẢN ĐỒ NHIỆT (HEATMAP) ĐỂ QUÉT TƯƠNG QUAN TỔNG THỂ', fontsize=14)
plt.show()
```

- Nhận xét:
- Profit (Lợi nhuận) và Product_Price (Giá bán) có tương quan rất yếu (0.28) => không cần tốn thời gian viết các câu lệnh truy vấn phức tạp để phân loại các mức giá đắt/rẻ nhằm tìm kiếm tỷ suất lợi nhuận.
- Tương quan giữa Profit và Units (Số lượng) là cao (0.56)=> phải ưu tiên tính tổng số lượng (SUM(Units)) trong SQL.

```python
#Tạo dim_date

#Tìm min,max date trong bảng sales
start_date = df_sales['Date'].min()
end_date = df_sales['Date'].max()

#Sinh ra lịch liên tục trong khoảng min max date
calendar_dates = pd.date_range(start = start_date, end = end_date)

#Dưa lịch vào dataframe
dim_date = pd.DataFrame({'Date':calendar_dates})

#Tạo các cột
dim_date['Day'] = dim_date['Date'].dt.day
dim_date['Month'] = dim_date['Date'].dt.month
dim_date['Quarter'] = dim_date['Date'].dt.quarter
dim_date['Year'] = dim_date['Date'].dt.year
dim_date['Day_of_Week'] = dim_date['Date'].dt.day_name()

dim_date.head(7)
```
